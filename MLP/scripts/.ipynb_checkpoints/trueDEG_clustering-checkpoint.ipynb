```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import csv
from sklearn.cluster import KMeans
```

# Load File

```python
trueDEGs_df = pd.read_csv(filepath_or_buffer='../results/biclass_trueDEGs_data.csv', index_col=0)
trueDEGs_df
```

# Try with Different K
- Inertia: The sum of squared distances between each sample and its assigned centroid.

```python
ks = range(1, 21)
inertias = []
for k in ks:
    # Create a KMeans instance with k clusters: model
    model = KMeans(n_clusters=k)
    
    # Fit model to samples
    model.fit(trueDEGs_df)
    
    # Append the inertia to the list of inertias
    inertias.append(model.inertia_)
    
# Plot ks vs inertias
plt.plot(ks, inertias, '-o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.xticks(ks)
plt.show()
```

# KMeans with K=2

```python
# Create a KMeans instance with 3 clusters: model
kmeans_model = KMeans(n_clusters=2)

# Fit model to data
kmeans_model.fit(trueDEGs_df)

# Determine the cluster labels of data
labels = kmeans_model.labels_

# Assign the columns of new_points: xs and ys
xs = trueDEGs_df.iloc[:,1]
ys = trueDEGs_df.iloc[:,2]

# Make a scatter plot of xs and ys, using labels to define the colors
plt.scatter(xs,ys,c=labels,alpha=0.5)

# Assign the cluster centers: centroids
centroids = kmeans_model.cluster_centers_

# Assign the columns of centroids: centroids_x, centroids_y
centroids_x = centroids[:,2]
centroids_y = centroids[:,3]

# Make a scatter plot of centroids_x and centroids_y
plt.scatter(centroids_x,centroids_y,marker='D',s=50)
plt.xlabel('CNV Mean')
plt.ylabel('Methylation Island')
plt.show()
```

```python
# Pick out genes within each cluster
cluster0_index = np.where(labels == 0)[0].tolist()
cluster1_index = np.where(labels == 1)[0].tolist()
cluter0_genes = trueDEGs_df.iloc[cluster0_index].index.tolist()
cluter1_genes = trueDEGs_df.iloc[cluster1_index].index.tolist()
```

```python
# Save the genes
with open("../results/cluter1_genes.csv","w") as f:
    wr = csv.writer(f, delimiter="\n")
    wr.writerow(cluter0_genes)
    
with open("../results/cluter2_genes.csv","w") as f:
    wr = csv.writer(f, delimiter="\n")
    wr.writerow(cluter1_genes)
```
