```python
import numpy as np
import pandas as pd
import feyn

from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from seaborn import clustermap
```

## Load the data
Note, the data has been preprocessed and missing values have been imputed.
It contains **2684 genes** and **7382 omics features**.

```python
data = pd.read_csv('/Users/yi/Documents/DTU/Master_Thesis/data/omics_samples_merged.csv')
# Delete first column
data = data.drop(data.columns[[0]], axis=1)

# Set row names
data = data.set_index('Gene.Symbol')

# Add binary target
data['gene_type'] = np.where(data['logFC'] < 0, 0, 1)
data = data.drop(columns = 'logFC', axis=1)

# # Delete CNV and Methy data
# data = data.drop(data.columns[0:1342], axis=1)

# Let's record the categorical data types in our dataset (note features will be treated as numerical by default).
stypes = {}
for f in data.columns:
    if data[f].dtype =='object':
        stypes[f] = 'c'
```

```python
data
```

## Split dataset into train and test set

```python
# Set random seed for reproducibility
random_seed = 42

# Define the target variable
target = "gene_type"

# Split
train, test = train_test_split(data, test_size = 0.33, stratify=data[target], random_state = random_seed)
```

## Train the QLattice

### Sample and fit models

This occurs in the following steps:

1. Sample models from the QLattice;
2. Fit the models by minimizing [BIC (Bayesian Information Criterion)](https://en.wikipedia.org/wiki/Bayesian_information_criterion);
3. Update the QLattice with the best models' structures;
4. Repeat the process;

This is all captured within the `auto_run` function

```python
# Connecting 
ql = feyn.QLattice(random_seed)

# Sample and fit models
models = ql.auto_run(
    data = train,
    kind = "classification",
    output_name = target,
    stypes = stypes,
    n_epochs = 30,
    max_complexity = 10
    )
```

# Visualization
### 1) Model display

```python
# Display all models
for model in models[0:5]:
    feyn.show_model(
        model = model,
        update_display = False
    )
```

### 2) Best model visualization

```python
best_model = models[0]

best_model.plot(
    data = train,
    compare_data = test
)
```

```python
best_model.plot_signal(train)
```

```python
inputs_list = []
for m in models:
    for i in m.inputs:
        inputs_list.append(i)
        
inputs_list = list(set(inputs_list))
```

```python
inputs_list
```

```python
corr_inputs = data[inputs_list].corr()
cluster_map_args = {'cmap': 'feyn-diverging', 'vmin': 0, 'vmax': 1}

clustermap(abs(corr_inputs), **cluster_map_args, figsize=(6, 6))
plt.savefig('../figures/classfication_clustermap.pdf', dpi=300, bbox_inches='tight')
```
